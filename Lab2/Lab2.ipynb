```python
stack = []

def push():
    element = input("Enter the element to push: ")
    stack.append(element)
    print(element, "pushed into stack")

def pop():
    if not stack:
        print("Stack is empty! Cannot pop.")
    else:
        removed = stack.pop()
        print(removed, "popped from stack")

def display():
    if not stack:
        print("Stack is empty!")
    else:
        print("Stack elements are:")
        for item in reversed(stack):
            print(item)

while True:
    print("\n--- Stack Menu ---")
    print("1. Push")
    print("2. Pop")
    print("3. Display")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == '1':
        push()
    elif choice == '2':
        pop()
    elif choice == '3':
        display()
    elif choice == '4':
        print("Exiting program.")
        break
    else:
        print("Invalid choice! Try again.")
```

```python
queue = []

def enqueue():
    element = input("Enter the element to enqueue: ")
    queue.append(element)
    print(element, "added to queue")

def dequeue():
    if not queue:
        print("Queue is empty! Cannot dequeue.")
    else:
        removed = queue.pop(0)
        print(removed, "removed from queue")

def display():
    if not queue:
        print("Queue is empty!")
    else:
        print("Queue elements are:")
        for item in queue:
            print(item)

while True:
    print("\n--- Queue Menu ---")
    print("1. Enqueue")
    print("2. Dequeue")
    print("3. Display")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == '1':
        enqueue()
    elif choice == '2':
        dequeue()
    elif choice == '3':
        display()
    elif choice == '4':
        print("Exiting program.") 
        break
    else:
        print("Invalid choice! Try again.")
```
