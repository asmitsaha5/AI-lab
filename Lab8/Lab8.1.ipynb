```python
import heapq
import time
```

```python
GOAL_STATE = (1, 2, 3,
              4, 5, 6,
              7, 8, 0)
```

```python
def misplaced_tiles(state):
    return sum(1 for i in range(9) 
               if state[i] != 0 and state[i] != GOAL_STATE[i])
```

```python
def manhattan_distance(state):
    distance = 0
    for i in range(9):
        if state[i] == 0:
            continue
        
        x1, y1 = divmod(i, 3)
        goal_index = GOAL_STATE.index(state[i])
        x2, y2 = divmod(goal_index, 3)
        
        distance += abs(x1 - x2) + abs(y1 - y2)
    
    return distance
```

```python
def astar(start_state, heuristic):
    open_list = []
    heapq.heappush(open_list, (heuristic(start_state), 0, start_state))

    visited = set()
    nodes_expanded = 0

    while open_list:
        f, g, current = heapq.heappop(open_list)

        if current == GOAL_STATE:
            return g, nodes_expanded

        if current in visited:
            continue

        visited.add(current)
        nodes_expanded += 1

        for neighbor in get_neighbors(current):
            if neighbor not in visited:
                heapq.heappush(open_list, 
                              (g + 1 + heuristic(neighbor),
                               g + 1,
                               neighbor))

    return None, nodes_expanded
```

```python
def greedy_bfs(start_state, heuristic):
    open_list = []
    heapq.heappush(open_list, (heuristic(start_state), start_state))

    visited = set()
    nodes_expanded = 0
    depth = 0

    while open_list:
        h, current = heapq.heappop(open_list)

        if current == GOAL_STATE:
            return depth, nodes_expanded

        if current in visited:
            continue

        visited.add(current)
        nodes_expanded += 1

        for neighbor in get_neighbors(current):
            if neighbor not in visited:
                heapq.heappush(open_list,
                              (heuristic(neighbor),
                               neighbor))

        depth += 1

    return None, nodes_expanded
```

```python
def get_neighbors(state):
    neighbors = []
    zero_index = state.index(0)
    x, y = divmod(zero_index, 3)

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dx, dy in moves:
        nx, ny = x + dx, y + dy
        
        if 0 <= nx < 3 and 0 <= ny < 3:
            new_index = nx * 3 + ny
            new_state = list(state)
            new_state[zero_index], new_state[new_index] = \
                new_state[new_index], new_state[zero_index]
            
            neighbors.append(tuple(new_state))

    return neighbors
```

```python
start_state = (1, 2, 3,
               4, 0, 6,
               7, 5, 8)

print("Start State:", start_state)
```

```python
start = time.time()
cost, nodes = astar(start_state, manhattan_distance)
end = time.time()

print("A* (Manhattan)")
print("Solution Cost:", cost)
print("Nodes Expanded:", nodes)
print("Time:", end - start)
```

```python
start = time.time()
cost, nodes = greedy_bfs(start_state, manhattan_distance)
end = time.time()

print("Greedy BFS (Manhattan)")
print("Approx Cost:", cost)
print("Nodes Expanded:", nodes)
print("Time:", end - start)
```

```python
start = time.time()
cost, nodes = astar(start_state, misplaced_tiles)
end = time.time()

print("A* (Misplaced Tiles)")
print("Solution Cost:", cost)
print("Nodes Expanded:", nodes)
print("Time:", end - start)
```
