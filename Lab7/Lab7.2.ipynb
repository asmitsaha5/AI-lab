```python
import random

GOAL = ((1,2,3),
        (4,5,6),
        (7,8,0))
```

```python
def misplaced_tiles(state):
    count = 0
    for i in range(3):
        for j in range(3):
            if state[i][j] != 0 and state[i][j] != GOAL[i][j]:
                count += 1
    return count
```

```python
def manhattan_distance(state):
    distance = 0
    for i in range(3):
        for j in range(3):
            value = state[i][j]
            if value != 0:
                goal_x = (value - 1) // 3
                goal_y = (value - 1) % 3
                distance += abs(i - goal_x) + abs(j - goal_y)
    return distance
```

```python
def get_neighbors(state):
    neighbors = []
    state_list = [list(row) for row in state]

    for i in range(3):
        for j in range(3):
            if state_list[i][j] == 0:
                x, y = i, j

    moves = [(-1,0),(1,0),(0,-1),(0,1)]

    for dx, dy in moves:
        nx, ny = x + dx, y + dy
        if 0 <= nx < 3 and 0 <= ny < 3:
            new_state = [row[:] for row in state_list]
            new_state[x][y], new_state[nx][ny] = new_state[nx][ny], new_state[x][y]
            neighbors.append(tuple(tuple(row) for row in new_state))

    return neighbors
```

```python
def print_state(state):
    for row in state:
        print(row)
    print()
```

```python
def simple_hill_climbing(start, heuristic):
    current = start
    path = [current]

    while True:
        current_h = heuristic(current)
        neighbors = get_neighbors(current)

        next_states = []
        for neighbor in neighbors:
            if heuristic(neighbor) <= current_h:
                next_states.append(neighbor)

        if not next_states:
            break

        # explore all equal/better states
        current = next_states[0]
        path.append(current)

        if current == GOAL:
            break

    return path
```

```python
def steepest_hill_climbing(start, heuristic):
    current = start
    path = [current]

    while True:
        neighbors = get_neighbors(current)
        best_h = heuristic(current)
        best_states = []

        for neighbor in neighbors:
            h = heuristic(neighbor)
            if h < best_h:
                best_h = h
                best_states = [neighbor]
            elif h == best_h:
                best_states.append(neighbor)

        if not best_states:
            break

        current = best_states[0]
        path.append(current)

        if current == GOAL:
            break

    return path
```

```python
def stochastic_hill_climbing(start, heuristic):
    current = start
    path = [current]

    while True:
        current_h = heuristic(current)
        neighbors = get_neighbors(current)

        better_neighbors = [n for n in neighbors if heuristic(n) <= current_h]

        if not better_neighbors:
            break

        current = random.choice(better_neighbors)
        path.append(current)

        if current == GOAL:
            break

    return path
```

```python
if __name__ == "__main__":

    start = ((1,2,3),
             (4,8,0),
             (7,6,5))

    print("Initial State:")
    print_state(start)

    print("Goal State:")
    print_state(GOAL)

    # -------- SIMPLE --------
    print("===== Simple Hill Climbing (H2) =====")
    path1 = simple_hill_climbing(start, manhattan_distance)
    for step, state in enumerate(path1):
        print("Step", step)
        print_state(state)

    # -------- STEEPEST --------
    print("===== Steepest-Ascent Hill Climbing (H2) =====")
    path2 = steepest_hill_climbing(start, manhattan_distance)
    for step, state in enumerate(path2):
        print("Step", step)
        print_state(state)

    # -------- STOCHASTIC --------
    print("===== Stochastic Hill Climbing (H2) =====")
    path3 = stochastic_hill_climbing(start, manhattan_distance)
    for step, state in enumerate(path3):
        print("Step", step)
        print_state(state)
```
