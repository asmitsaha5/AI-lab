```python
import heapq
import copy

GOAL_STATE = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]

INITIAL_STATE = [
    [1, 2, 3],
    [4, 0, 6],
    [7, 5, 8]
]


def print_state(state):
    for row in state:
        print(row)
    print("------")

def find_blank(state):
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return i, j

def is_goal(state):
    return state == GOAL_STATE




def h1(state):
    count = 0
    for i in range(3):
        for j in range(3):
            if state[i][j] != 0 and state[i][j] != GOAL_STATE[i][j]:
                count += 1
    return count


def h2(state):
    distance = 0
    for i in range(3):
        for j in range(3):
            val = state[i][j]
            if val != 0:
                goal_x = (val - 1) // 3
                goal_y = (val - 1) % 3
                distance += abs(i - goal_x) + abs(j - goal_y)
    return distance



def get_neighbors(state):
    x, y = find_blank(state)
    moves = [(-1,0),(1,0),(0,-1),(0,1)]
    neighbors = []

    for dx, dy in moves:
        nx, ny = x + dx, y + dy
        if 0 <= nx < 3 and 0 <= ny < 3:
            new_state = copy.deepcopy(state)
            new_state[x][y], new_state[nx][ny] = new_state[nx][ny], new_state[x][y]
            neighbors.append(new_state)

    return neighbors



def astar(initial, heuristic):
    pq = []
    visited = set()

    heapq.heappush(pq, (heuristic(initial), 0, initial, []))

    while pq:
        f, g, state, path = heapq.heappop(pq)

        state_tuple = tuple(map(tuple, state))
        if state_tuple in visited:
            continue
        visited.add(state_tuple)

        path = path + [state]

        if is_goal(state):
            print("\nSolution Found (A*)")
            for step in path:
                print_state(step)
            print("Total steps:", len(path)-1)
            print("Nodes explored:", len(visited))
            return

        for neighbor in get_neighbors(state):
            heapq.heappush(pq, (g + 1 + heuristic(neighbor), g + 1, neighbor, path))

    print("No solution found")



def hill_climbing(initial):
    current = initial
    path = [current]

    while True:
        neighbors = get_neighbors(current)
        current_h = h2(current)

        next_state = None
        best_h = current_h

        for neighbor in neighbors:
            h_val = h2(neighbor)
            if h_val < best_h:
                best_h = h_val
                next_state = neighbor

        if next_state is None:
            break

        current = next_state
        path.append(current)

        if is_goal(current):
            print("\nSolution Found (Hill Climbing)")
            for step in path:
                print_state(step)
            print("Steps:", len(path)-1)
            return

    print("\nHill Climbing stuck (Local Optimum)")
    for step in path:
        print_state(step)


print("A* using H1 (Misplaced Tiles)")
astar(INITIAL_STATE, h1)

print("\nA* using H2 (Manhattan Distance)")
astar(INITIAL_STATE, h2)

print("\nHill Climbing (using H2)")
hill_climbing(INITIAL_STATE)
```
