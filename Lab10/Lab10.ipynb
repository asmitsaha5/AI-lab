```python
import math

# Board representation
board = [
    ['_', '_', '_'],
    ['_', '_', '_'],
    ['_', '_', '_']
]

player = 'X'   # AI
opponent = 'O' # Human


# Function to check if there are moves left
def isMovesLeft(board):
    for i in range(3):
        for j in range(3):
            if board[i][j] == '_':
                return True
    return False


# Evaluation function
def evaluate(board):

    # Rows
    for row in range(3):
        if board[row][0] == board[row][1] == board[row][2]:
            if board[row][0] == player:
                return 10
            elif board[row][0] == opponent:
                return -10

    # Columns
    for col in range(3):
        if board[0][col] == board[1][col] == board[2][col]:
            if board[0][col] == player:
                return 10
            elif board[0][col] == opponent:
                return -10

    # Diagonals
    if board[0][0] == board[1][1] == board[2][2]:
        if board[0][0] == player:
            return 10
        elif board[0][0] == opponent:
            return -10

    if board[0][2] == board[1][1] == board[2][0]:
        if board[0][2] == player:
            return 10
        elif board[0][2] == opponent:
            return -10

    return 0


# Minimax with Alpha-Beta Pruning
def minimax(board, depth, isMax, alpha, beta):

    score = evaluate(board)

    # If AI wins
    if score == 10:
        return score - depth   # faster win preferred

    # If opponent wins
    if score == -10:
        return score + depth   # slower loss preferred

    # Draw
    if not isMovesLeft(board):
        return 0

    # Maximizer (AI)
    if isMax:
        best = -math.inf

        for i in range(3):
            for j in range(3):

                if board[i][j] == '_':
                    board[i][j] = player

                    value = minimax(board, depth+1, False, alpha, beta)

                    best = max(best, value)
                    alpha = max(alpha, best)

                    board[i][j] = '_'

                    if beta <= alpha:
                        break

        return best

    # Minimizer (Opponent)
    else:
        best = math.inf

        for i in range(3):
            for j in range(3):

                if board[i][j] == '_':
                    board[i][j] = opponent

                    value = minimax(board, depth+1, True, alpha, beta)

                    best = min(best, value)
                    beta = min(beta, best)

                    board[i][j] = '_'

                    if beta <= alpha:
                        break

        return best


# Find the best move
def findBestMove(board):

    bestVal = -math.inf
    bestMove = (-1, -1)

    for i in range(3):
        for j in range(3):

            if board[i][j] == '_':
                board[i][j] = player

                moveVal = minimax(board, 0, False, -math.inf, math.inf)

                board[i][j] = '_'

                if moveVal > bestVal:
                    bestMove = (i, j)
                    bestVal = moveVal

    return bestMove


# Display board
def printBoard(board):
    for row in board:
        print(" ".join(row))
    print()


# Simple game loop
while isMovesLeft(board):

    printBoard(board)

    # Human move
    r = int(input("Enter row (0-2): "))
    c = int(input("Enter column (0-2): "))

    if board[r][c] != '_':
        print("Invalid move")
        continue

    board[r][c] = opponent

    if evaluate(board) == -10:
        printBoard(board)
        print("You win!")
        break

    if not isMovesLeft(board):
        print("Draw!")
        break

    # AI move
    move = findBestMove(board)
    board[move[0]][move[1]] = player
    print("AI played:", move)

    if evaluate(board) == 10:
        printBoard(board)
        print("AI wins!")
        break
```
